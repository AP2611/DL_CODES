```python
import tensorflow as tf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
import seaborn as sns
```

```python
data = pd.read_csv('ecg.csv')
```

```python
data.head()
```

```python
x = data.iloc[:, :-1]
y = data.iloc[:, -1]
```

```python
x_normal = x[y==0]
x_abnormal = x[y==1]
```

```python
x_train,x_test_normal = train_test_split(
    x_normal,
    test_size=42,
    random_state = 42
)
```

```python
x_test = pd.concat([x_test_normal,x_abnormal])
y_test = np.concatenate([
    np.zeros(len(x_test_normal)),
    np.ones(len(x_abnormal))
])
```

```python
scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
n = x_train.shape[1]
model = Sequential()
```

```python
model.add(Dense(32, activation='relu', input_shape=(n,)))
model.add(Dense(16, activation='relu'))
model.add(Dense(8, activation='relu'))
```

```python
model.add(Dense(16,activation = 'relu'))
model.add(Dense(32,activation = 'relu'))
model.add(Dense(n,activation = 'linear'))
```

```python
model.compile(
    optimizer = 'adam',
    loss = 'mse',
    metrics = ['mae']
)
```

```python
model.summary()
```

```python
model.fit(
    x_train,x_train,
    epochs = 30,
    batch_size = 32,
    validation_split = 0.2
)
```

```python
reconstructed = model.predict(x_test)
```

```python
error = np.mean(
    np.square(x_test - reconstructed),
    axis = 1
)
```

```python
train_reconstructed = model.predict(x_train)
train_error = np.mean(
    np.square(x_train - train_reconstructed),
    axis = 1
)
```

```python
threshold = np.percentile(train_error,90)
print("Threshold = ",threshold)
```

```python
predictions = (error > threshold).astype(int)
for i in range(10):
    if predictions[i] == 1:
        print("Abnormal")
    else:
        print("Normal")
        
```

```python
print(classification_report(y_test, predictions))
```

```python
print("Accuracy:", accuracy_score(y_test, predictions) * 100)

print("Precision:", precision_score(y_test, predictions) * 100)

print("Recall:", recall_score(y_test, predictions) * 100)

print("F1 Score:", f1_score(y_test, predictions) * 100)
```
